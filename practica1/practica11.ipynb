### **2. Cuestionario previo**

* **¿Qué es Jupyter?** 

Es una aplicación interactiva que permite programar y ejecutar código por bloques, además de poder implementar texto con formato, ecuaciones, visualizaciones y demás elementos interactivos. 

* **¿Qué es un archivo URDF?**

Un archivo Unified Robot Description Format es un archivo en formato XML utilizado en robótica para describir la estructura de un robot, incluyendo eslabones, juntas, sensores y dimensiones físicas. 

* **¿Qué es Git y GitHub?** 

Git es un sistema de control de versiones que permite guardar cambios en proyectos y trabajar en equipo, mientras que Github es una plataforma en línea donde se almacenan repositorios de Git para compartir, colaborar y respaldar proyectos.

* **¿Qué es RViz?**

RViz es una herramienta de visualización de ROS que permite ver robots, sensores, mapas, trayectorias y movimientos en 3D para simulación y depuración de sistemas robóticos.</div>


### **3. Desarrollo**

#### **3.1. Creación de un modelo URDF**

El modelo solicitado a desarrollar es el siguiente:

En el cual se proponen las dimensiones a conveniencia, con el único requisito de no tomar medidas mayores a 0.5 [m]. 

A continuación se describe cada una de las partes del archvio para la creación del URDF.

#### **Asignación de Nombre del Robot**

Primero, se colocó el nombre de "robot" a nuestro archivo URDF.

```python
<robot name="robot" xmlns:xacro="http://www.ros.org/wiki/xacro">
```

#### **Creación y descripción de los eslabones**

En esta sección se definen los distintos eslabones que conforman la estructura física del robot serial. Cada bloque <link> representa un cuerpo rígido del manipulador, especificando su geometría, dimensiones, posición visual y color dentro del entorno RViz.

* **Base principal del robot**

Primero, se crea la base inferior del robot mediante un cilindro de radio grande y poca altura, el cual funciona como soporte principal de toda la estructura.

```python
  <link name="base_link1">
    <visual>
      <geometry>
        <cylinder radius="0.15" length="0.05"/>
      </geometry>
      <origin xyz="0 0 -0.025"/>
      <material name="black">
        <color rgba="0 0 0 0.5"/>
      </material>
    </visual>
  </link>
```

* **Columna vertical de la base**

Posteriormente, se agrega un segundo eslabón cilíndrico vertical el cual sigue siendo parte de la base.

```python
  <link name="base_link2">
    <visual>
      <geometry>
        <cylinder radius="0.05" length="0.3"/>
      </geometry>
      <origin xyz="0 0 0.15"/>
      <material name="cyan">
        <color rgba="0 0 0 0.5"/>
      </material>
    </visual>
  </link>
```

* **Eslabón del hombro**

Después, se modela el eslabón correspondiente al hombro del robot mediante una geometría tipo prisma rectangular. Este eslabón representa el primer brazo móvil del sistema.

```python
<link name="shoulder_link">
    <visual>
      <geometry>
        <box size="0.3 0.05 0.05"/>
      </geometry>
      <origin xyz="0.1 0 0"/>
      <material name="cyan">
        <color rgba="0 1 1 1"/>
      </material>
    </visual>
  </link>
```

* **Primer brazo del manipulador**

Se define el segundo brazo del robot utilizando otro prisma rectangular, el cual extiende el alcance del manipulador.

```python
  <link name="arm_link1">
    <visual>
      <geometry>
        <box size="0.3 0.05 0.05"/>
      </geometry>
      <origin xyz="0.15 0 0"/>
      <material name="magenta">
        <color rgba="1 0 1 1"/>
      </material>
    </visual>
  </link>
```

* **Soporte intermedio del brazo**

En esta parte se crea un eslabón cilíndrico auxiliar que sirve como conexión entre el brazo principal y el sistema prismático final.

```python
  <link name="arm_link2">
    <visual>
      <geometry>
        <cylinder radius="0.025" length="0.1"/>
      </geometry>
      <origin xyz="0.275 0 -0.075"/>
      <material name="magenta">
        <color rgba="1 0 1 1"/>
      </material>
    </visual>
  </link>
```

* **Eslabón prismático final**

Por último se modela el efector prismático del robot mediante un cilindro vertical que representa el movimiento lineal del sistema.

```python
  <link name="forearm_link">
    <visual>
      <geometry>
        <cylinder radius="0.02" length="0.1"/>
      </geometry>
      <origin xyz="0 0 0"/>
      <material name="yellow">
        <color rgba="1 1 0 1"/>
      </material>
    </visual>
  </link>
```

#### **Definición de las juntas del robot**

Una vez definidos los eslabones, se establecen las juntas encargadas de conectar cada componente y permitir el movimiento del robot. Estas juntas determinan tanto el tipo de movimiento como las relaciones cinemáticas entre eslabones.

* **Junta fija de la base**

Primero se conecta la base inferior con la columna vertical mediante una junta fija, la cual no permite movimiento relativo entre ambos eslabones.

```python
  <joint name="base_fixed_joint" type="fixed"> <!--tipo de junta: revolute, prismatic, fixed-->
    <parent link="base_link1"/>
    <child link="base_link2"/>
    <origin rpy="0 0 0" xyz="0 0 0"/> <!--roll, pitch, yaw-->
  </joint>
```

* **Junta rotacional del hombro**

Luego se implementa una junta revoluta en el hombro del robot. Esta junta permite la rotación alrededor del eje z, lo que origina el primer grado de libertad angular del manipulador.

```python
  <joint name="shoulder_joint" type="revolute"> <!--tipo de junta: revolute, prismatic, fixed-->
    <parent link="base_link2"/>
    <child link="shoulder_link"/>
    <axis xyz="0 0 1"/>
    <limit lower="-3.14" upper ="3.14" effort="10.0" velocity="10.0"/>
    <origin rpy="0 0 0" xyz="0 0 0.325"/> <!--roll, pitch, yaw-->
  </joint>
```

* **Junta rotacional del brazo**

Después se agrega una segunda junta revoluta que conecta el hombro con el brazo principal. Esta articulación permite el movimiento angular del segundo eslabón del robot.

```python
  <joint name="arm_joint" type="revolute"> <!--tipo de junta: revolute, prismatic, fixed-->
    <parent link="shoulder_link"/>
    <child link="arm_link1"/>
    <axis xyz="0 0 1"/>
    <limit lower="-3.14" upper ="3.14" effort="10.0" velocity="10.0"/>
    <origin xyz="0.225 0 0.05" rpy="0 0 0"/>
  </joint>
```

* **Junta fija intermedia**

En esta parte se conecta el soporte cilíndrico auxiliar a través de una junta fija, manteniendo una unión rígida entre ambos.

```python
  <joint name="arm_fixed_joint" type="fixed"> <!--tipo de junta: revolute, prismatic, fixed-->
    <parent link="arm_link1"/>
    <child link="arm_link2"/>
    <origin rpy="0 0 0" xyz="0 0 0"/> <!--roll, pitch, yaw-->
  </joint>
```

* **Junta de tipo prismática final**

Por último se implementa una junta prismática encargada del movimiento lineal del efector final. Esta junta permite el desplazamiento sobre el eje z, simulando la extensión y retracción del manipulador.

```python
  <joint name="forearm_joint" type="prismatic"> <!--tipo de junta: revolute, prismatic, fixed-->
    <parent link="arm_link2"/>
    <child link="forearm_link"/>
    <axis xyz="0 0 1"/>
    <limit lower="-0.05" upper ="0.05" effort="10.0" velocity="10.0"/>
    <origin rpy="0 0 0" xyz="0.275 0 -0.125"/>
  </joint>
```

### **4. Análisis de resultados**

<div style="text-align: justify;">

* **¿Qué utilidad tiene el uso de un repositorio en línea para código?** 

Es de gran utilidad ya que un repositorio en línea permite almacenar, respaldar y compartir código de manera organizada; además de facilitar el trabajo colaborativo entre varios, el control de versiones, la recuperación de cambios anteriores y la sincronización del proyecto desde distintos equipos.

* **¿Cuál es la convención de ángulos que utilizan los archivos URDF para los ángulos de las juntas respecto al eslabón padre?**

Utilizan ángulos expresados en radianes y siguen la convención Roll-Pitch-Yaw (RPY) para describir orientaciones respecto al eslabón padre. Estas rotaciones se aplican normalmente alrededor de los ejes x, y y z respectivamente.

* **¿Qué utilidad tiene describir un robot en un archivo URDF?** 

La descripción de un robot en un archivo URDF permite representar su estructura física y cinemática dentro de ROS. Mediante este archivo se pueden definir eslabones, juntas, dimensiones, geometrías, sensores y relaciones de movimiento, facilitando la visualización, simulación y control del robot en herramientas como RViz y Gazebo.
