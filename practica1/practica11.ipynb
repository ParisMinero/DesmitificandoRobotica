### **2. Cuestionario previo**

* **¿Qué es Jupyter?** 

Es una aplicación interactiva que permite programar y ejecutar código por bloques, además de poder implementar texto con formato, ecuaciones, visualizaciones y demás elementos interactivos. 

* **¿Qué es un archivo URDF?**

Un archivo Unified Robot Description Format es un archivo en formato XML utilizado en robótica para describir la estructura de un robot, incluyendo eslabones, juntas, sensores y dimensiones físicas. 

* **¿Qué es Git y GitHub?** 

Git es un sistema de control de versiones que permite guardar cambios en proyectos y trabajar en equipo, mientras que Github es una plataforma en línea donde se almacenan repositorios de Git para compartir, colaborar y respaldar proyectos.

* **¿Qué es RViz?**

RViz es una herramienta de visualización de ROS que permite ver robots, sensores, mapas, trayectorias y movimientos en 3D para simulación y depuración de sistemas robóticos.</div>
